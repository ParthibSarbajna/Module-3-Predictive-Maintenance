```python
# ================================================================
# MODULE 3 - AI/ML BASED PREDICTIVE MAINTENANCE & VEHICLE HEALTH
# MONITORING SYSTEM
#
# Wipro Capstone Project - Module 3 (Connectivity)
#
# REAL DATASET:
# jskswamy/predictive-maintenance-data
#
# Models:
# 1. Random Forest
# 2. XGBoost
# 3. SHAP Explainable AI
#
# Outputs:
# - Failure Prediction
# - Failure Probability
# - Vehicle Health Score
# - Maintenance Status
# - Maintenance Alert
# - Major Contributing Factors
# - Model Comparison
# - Confusion Matrix
# - ROC Curve
# - Feature Importance
# - SHAP Explainability
# ================================================================


# ================================================================
# 1. INSTALL REQUIRED LIBRARIES
# ================================================================

!pip -q install datasets xgboost shap joblib


# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import os
import json
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from datasets import load_dataset

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

from xgboost import XGBClassifier

import shap

from IPython.display import display, Markdown


# ================================================================
# 3. PROJECT INFORMATION
# ================================================================

print("=" * 75)
print(" AI/ML-BASED PREDICTIVE MAINTENANCE AND VEHICLE HEALTH MONITORING")
print("=" * 75)
print()
print("Project: Wipro Capstone - Module 3")
print("Dataset: Real Engine Sensor Predictive Maintenance Dataset")
print("Models: Random Forest + XGBoost")
print("Explainability: SHAP")
print()


# ================================================================
# 4. DOWNLOAD REAL DATASET FROM HUGGING FACE
# ================================================================

print("Downloading real predictive-maintenance dataset...")
print()

dataset = load_dataset("jskswamy/predictive-maintenance-data")

print("Dataset downloaded successfully!")
print()
print(dataset)
print()


# ================================================================
# 5. COMBINE AVAILABLE DATASET SPLITS
# ================================================================

frames = []

for split_name in dataset.keys():
    temp_df = dataset[split_name].to_pandas()
    temp_df["source_split"] = split_name
    frames.append(temp_df)

df = pd.concat(frames, ignore_index=True)

print("Combined dataset shape:", df.shape)
print()
print("Available columns:")
print(df.columns.tolist())
print()


# ================================================================
# 6. DISPLAY RAW DATA
# ================================================================

print("=" * 75)
print("FIRST 10 RECORDS")
print("=" * 75)

display(df.head(10))

print()
print("Dataset information:")
print()

df.info()


# ================================================================
# 7. DATA CLEANING
# ================================================================

print()
print("=" * 75)
print("DATA CLEANING")
print("=" * 75)

# Remove duplicate rows
before_duplicates = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after_duplicates = len(df)

print("Duplicate rows removed:", before_duplicates - after_duplicates)

# Remove rows with missing values in required columns
required_columns = [
    "Engine RPM",
    "Lub Oil Pressure",
    "Fuel Pressure",
    "Coolant Pressure",
    "Lub Oil Temp",
    "Coolant Temp",
    "Engine Condition"
]

missing_required = [col for col in required_columns if col not in df.columns]

if len(missing_required) > 0:
    raise ValueError(
        f"Required columns not found: {missing_required}\n"
        f"Available columns: {df.columns.tolist()}"
    )

df = df.dropna(subset=required_columns).reset_index(drop=True)

print("Final dataset shape after cleaning:", df.shape)

print()
print("Missing values:")
print(df[required_columns].isnull().sum())


# ================================================================
# 8. USE ONLY RAW SENSOR FEATURES
# ================================================================
#
# The dataset also contains several engineered features.
# For a clean project demonstration, we deliberately use the
# original six sensor readings instead of using pre-engineered
# variables that could make the model less transparent.
# ================================================================

feature_columns = [
    "Engine RPM",
    "Lub Oil Pressure",
    "Fuel Pressure",
    "Coolant Pressure",
    "Lub Oil Temp",
    "Coolant Temp"
]

target_column = "Engine Condition"

X = df[feature_columns].copy()
y = df[target_column].astype(int).copy()


# ================================================================
# 9. TARGET INFORMATION
# ================================================================

print()
print("=" * 75)
print("TARGET DISTRIBUTION")
print("=" * 75)

target_counts = y.value_counts().sort_index()

print("0 = Normal / Healthy")
print("1 = Maintenance Required / Faulty")
print()

print(target_counts)

print()

target_percent = y.value_counts(normalize=True).sort_index() * 100

print("Target percentages:")
print(target_percent.round(2))


# ================================================================
# 10. BASIC STATISTICS
# ================================================================

print()
print("=" * 75)
print("SENSOR STATISTICS")
print("=" * 75)

display(X.describe().round(3))


# ================================================================
# 11. EXPLORATORY DATA ANALYSIS
# ================================================================

print()
print("=" * 75)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 75)


# Target distribution
plt.figure(figsize=(7, 5))

sns.countplot(
    x=y,
    hue=y,
    legend=False
)

plt.title("Engine Condition Distribution")
plt.xlabel("Engine Condition")
plt.ylabel("Number of Records")
plt.xticks(
    [0, 1],
    ["Normal / Healthy", "Maintenance Required"]
)

plt.tight_layout()
plt.show()


# Sensor distributions
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for ax, column in zip(axes.flatten(), feature_columns):

    sns.histplot(
        data=df,
        x=column,
        hue=target_column,
        kde=True,
        element="step",
        ax=ax
    )

    ax.set_title(f"{column} Distribution")

plt.tight_layout()
plt.show()


# ================================================================
# 12. CORRELATION MATRIX
# ================================================================

plt.figure(figsize=(10, 7))

correlation_data = df[feature_columns + [target_column]].corr()

sns.heatmap(
    correlation_data,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Sensor and Engine Condition Correlation Matrix")

plt.tight_layout()
plt.show()


# ================================================================
# 13. TRAIN / TEST SPLIT
# ================================================================

print()
print("=" * 75)
print("TRAIN / TEST SPLIT")
print("=" * 75)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print()
print("Training class distribution:")
print(y_train.value_counts())

print()
print("Testing class distribution:")
print(y_test.value_counts())


# ================================================================
# 14. RANDOM FOREST MODEL
# ================================================================

print()
print("=" * 75)
print("TRAINING RANDOM FOREST")
print("=" * 75)

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=15,
    min_samples_split=5,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)

rf_probabilities = rf_model.predict_proba(X_test)[:, 1]


# ================================================================
# 15. RANDOM FOREST METRICS
# ================================================================

rf_accuracy = accuracy_score(y_test, rf_predictions)
rf_precision = precision_score(y_test, rf_predictions)
rf_recall = recall_score(y_test, rf_predictions)
rf_f1 = f1_score(y_test, rf_predictions)
rf_auc = roc_auc_score(y_test, rf_probabilities)

rf_metrics = {
    "Accuracy": rf_accuracy,
    "Precision": rf_precision,
    "Recall": rf_recall,
    "F1 Score": rf_f1,
    "ROC-AUC": rf_auc
}

print()
print("RANDOM FOREST RESULTS")
print("-" * 50)

for metric, value in rf_metrics.items():
    print(f"{metric:<15}: {value:.4f}")


# ================================================================
# 16. XGBOOST MODEL
# ================================================================

print()
print("=" * 75)
print("TRAINING XGBOOST")
print("=" * 75)

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.85,
    colsample_bytree=0.85,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train, y_train)

xgb_predictions = xgb_model.predict(X_test)

xgb_probabilities = xgb_model.predict_proba(X_test)[:, 1]


# ================================================================
# 17. XGBOOST METRICS
# ================================================================

xgb_accuracy = accuracy_score(y_test, xgb_predictions)
xgb_precision = precision_score(y_test, xgb_predictions)
xgb_recall = recall_score(y_test, xgb_predictions)
xgb_f1 = f1_score(y_test, xgb_predictions)
xgb_auc = roc_auc_score(y_test, xgb_probabilities)

xgb_metrics = {
    "Accuracy": xgb_accuracy,
    "Precision": xgb_precision,
    "Recall": xgb_recall,
    "F1 Score": xgb_f1,
    "ROC-AUC": xgb_auc
}

print()
print("XGBOOST RESULTS")
print("-" * 50)

for metric, value in xgb_metrics.items():
    print(f"{metric:<15}: {value:.4f}")


# ================================================================
# 18. MODEL COMPARISON
# ================================================================

comparison = pd.DataFrame(
    [rf_metrics, xgb_metrics],
    index=["Random Forest", "XGBoost"]
)

print()
print("=" * 75)
print("MODEL COMPARISON")
print("=" * 75)

display(comparison.round(4))


# ================================================================
# 19. MODEL COMPARISON GRAPH
# ================================================================

comparison.plot(
    kind="bar",
    figsize=(13, 7)
)

plt.title("Random Forest vs XGBoost Performance")
plt.xlabel("Model")
plt.ylabel("Score")
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.show()


# ================================================================
# 20. ROC CURVE
# ================================================================

rf_fpr, rf_tpr, _ = roc_curve(
    y_test,
    rf_probabilities
)

xgb_fpr, xgb_tpr, _ = roc_curve(
    y_test,
    xgb_probabilities
)

plt.figure(figsize=(9, 7))

plt.plot(
    rf_fpr,
    rf_tpr,
    linewidth=2,
    label=f"Random Forest (AUC = {rf_auc:.3f})"
)

plt.plot(
    xgb_fpr,
    xgb_tpr,
    linewidth=2,
    label=f"XGBoost (AUC = {xgb_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    "--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve - Predictive Maintenance Models")

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()


# ================================================================
# 21. CONFUSION MATRIX - XGBOOST
# ================================================================

cm = confusion_matrix(
    y_test,
    xgb_predictions
)

plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Normal",
        "Maintenance Required"
    ],
    yticklabels=[
        "Normal",
        "Maintenance Required"
    ]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.title("XGBoost Confusion Matrix")

plt.tight_layout()
plt.show()


# ================================================================
# 22. CLASSIFICATION REPORT
# ================================================================

print()
print("=" * 75)
print("XGBOOST CLASSIFICATION REPORT")
print("=" * 75)

print(
    classification_report(
        y_test,
        xgb_predictions,
        target_names=[
            "Normal / Healthy",
            "Maintenance Required"
        ]
    )
)


# ================================================================
# 23. RANDOM FOREST FEATURE IMPORTANCE
# ================================================================

rf_importance = pd.DataFrame({
    "Feature": feature_columns,
    "Importance": rf_model.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

print()
print("=" * 75)
print("RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 75)

display(rf_importance.round(4))


plt.figure(figsize=(10, 6))

sns.barplot(
    data=rf_importance,
    x="Importance",
    y="Feature"
)

plt.title("Random Forest Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Sensor Parameter")

plt.tight_layout()
plt.show()


# ================================================================
# 24. XGBOOST FEATURE IMPORTANCE
# ================================================================

xgb_importance = pd.DataFrame({
    "Feature": feature_columns,
    "Importance": xgb_model.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

print()
print("=" * 75)
print("XGBOOST FEATURE IMPORTANCE")
print("=" * 75)

display(xgb_importance.round(4))


plt.figure(figsize=(10, 6))

sns.barplot(
    data=xgb_importance,
    x="Importance",
    y="Feature"
)

plt.title("XGBoost Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Sensor Parameter")

plt.tight_layout()
plt.show()


# ================================================================
# 25. SHAP EXPLAINABLE AI
# ================================================================

print()
print("=" * 75)
print("SHAP EXPLAINABLE AI")
print("=" * 75)

# Use a sample to keep SHAP execution fast in Colab
shap_sample_size = min(1000, len(X_test))

X_shap = X_test.sample(
    shap_sample_size,
    random_state=42
)

explainer = shap.TreeExplainer(xgb_model)

shap_values = explainer.shap_values(X_shap)

print("SHAP explanation generated successfully.")


# SHAP summary
plt.figure()

shap.summary_plot(
    shap_values,
    X_shap,
    show=False
)

plt.title("SHAP Feature Impact - XGBoost")

plt.tight_layout()
plt.show()


# ================================================================
# 26. MEAN ABSOLUTE SHAP IMPORTANCE
# ================================================================

mean_abs_shap = np.abs(shap_values).mean(axis=0)

shap_importance = pd.DataFrame({
    "Feature": feature_columns,
    "Mean Absolute SHAP": mean_abs_shap
}).sort_values(
    "Mean Absolute SHAP",
    ascending=False
)

print()
print("SHAP FEATURE IMPORTANCE")
print("-" * 50)

display(shap_importance.round(5))


# ================================================================
# 27. SELECT DEPLOYED MODEL
# ================================================================
#
# We use XGBoost as the prediction engine for the demonstration.
# The selection is based on ROC-AUC from the evaluated test set.
# ================================================================

if xgb_auc >= rf_auc:
    deployed_model = xgb_model
    deployed_model_name = "XGBoost"
    deployed_probabilities = xgb_probabilities
else:
    deployed_model = rf_model
    deployed_model_name = "Random Forest"
    deployed_probabilities = rf_probabilities

print()
print("=" * 75)
print("SELECTED PREDICTION MODEL")
print("=" * 75)

print("Model used for vehicle health monitoring:", deployed_model_name)


# ================================================================
# 28. VEHICLE HEALTH MONITORING FUNCTION
# ================================================================

def vehicle_health_monitor(
    engine_rpm,
    lub_oil_pressure,
    fuel_pressure,
    coolant_pressure,
    lub_oil_temp,
    coolant_temp
):

    vehicle_data = pd.DataFrame([{
        "Engine RPM": engine_rpm,
        "Lub Oil Pressure": lub_oil_pressure,
        "Fuel Pressure": fuel_pressure,
        "Coolant Pressure": coolant_pressure,
        "Lub Oil Temp": lub_oil_temp,
        "Coolant Temp": coolant_temp
    }])

    failure_probability = float(
        deployed_model.predict_proba(vehicle_data)[0][1]
    )

    predicted_condition = int(
        deployed_model.predict(vehicle_data)[0]
    )

    # Health score
    health_score = 100 * (1 - failure_probability)

    health_score = max(
        0,
        min(100, health_score)
    )

    # Status classification
    if failure_probability < 0.25:

        status = "NORMAL"
        maintenance_alert = (
            "Vehicle is operating within a relatively safe "
            "predicted condition. Continue routine monitoring."
        )

    elif failure_probability < 0.50:

        status = "MONITOR"
        maintenance_alert = (
            "Abnormal operating patterns detected. "
            "Increase monitoring and schedule inspection."
        )

    elif failure_probability < 0.75:

        status = "WARNING"
        maintenance_alert = (
            "High maintenance risk detected. "
            "Maintenance inspection is recommended soon."
        )

    else:

        status = "CRITICAL"
        maintenance_alert = (
            "Very high predicted maintenance risk. "
            "Immediate inspection is recommended."
        )


    # ------------------------------------------------------------
    # Identify contributing sensor conditions
    # ------------------------------------------------------------

    factors = []

    if lub_oil_pressure < 3.0:
        factors.append("Low lubrication oil pressure")

    if fuel_pressure < 5.0:
        factors.append("Low fuel pressure")

    if coolant_pressure < 2.0:
        factors.append("Low coolant pressure")

    if lub_oil_temp > 82:
        factors.append("High lubrication oil temperature")

    if coolant_temp > 82:
        factors.append("High coolant temperature")

    if engine_rpm > 2500:
        factors.append("High engine RPM")

    if len(factors) == 0:
        factors.append(
            "No major predefined sensor threshold violation"
        )


    return {
        "Failure Probability": failure_probability,
        "Health Score": health_score,
        "Predicted Condition": predicted_condition,
        "Status": status,
        "Maintenance Alert": maintenance_alert,
        "Contributing Factors": factors
    }


# ================================================================
# 29. TEST VEHICLE / SAMPLE VEHICLE
# ================================================================

# This sample intentionally contains several abnormal readings
# so that the complete alerting pipeline can be demonstrated.

sample_vehicle = {
    "engine_rpm": 3200,
    "lub_oil_pressure": 2.4,
    "fuel_pressure": 4.2,
    "coolant_pressure": 1.6,
    "lub_oil_temp": 86,
    "coolant_temp": 89
}

vehicle_report = vehicle_health_monitor(
    **sample_vehicle
)


# ================================================================
# 30. DISPLAY VEHICLE HEALTH REPORT
# ================================================================

print()
print("=" * 75)
print(" VEHICLE HEALTH MONITORING REPORT")
print("=" * 75)

print()

print(
    f"Vehicle Health Score     : "
    f"{vehicle_report['Health Score']:.2f}%"
)

print(
    f"Failure Probability      : "
    f"{vehicle_report['Failure Probability'] * 100:.2f}%"
)

print(
    f"Predicted Condition      : "
    f"{'MAINTENANCE REQUIRED' if vehicle_report['Predicted Condition'] == 1 else 'NORMAL'}"
)

print(
    f"Vehicle Status           : "
    f"{vehicle_report['Status']}"
)

print()

print("Maintenance Alert:")
print(vehicle_report["Maintenance Alert"])

print()

print("Major Contributing Factors:")

for i, factor in enumerate(
    vehicle_report["Contributing Factors"],
    1
):

    print(f"{i}. {factor}")

print()
print("=" * 75)


# ================================================================
# 31. VEHICLE HEALTH VISUALIZATION
# ================================================================

health_score = vehicle_report["Health Score"]

plt.figure(figsize=(10, 3))

plt.barh(
    ["Vehicle Health"],
    [health_score]
)

plt.xlim(0, 100)

plt.xlabel("Health Score (%)")

plt.title(
    f"Vehicle Health Score: {health_score:.2f}% "
    f"({vehicle_report['Status']})"
)

plt.grid(
    axis="x",
    alpha=0.25
)

plt.tight_layout()
plt.show()


# ================================================================
# 32. SAMPLE SENSOR DASHBOARD
# ================================================================

sensor_values = pd.DataFrame({
    "Sensor Parameter": [
        "Engine RPM",
        "Lub Oil Pressure",
        "Fuel Pressure",
        "Coolant Pressure",
        "Lub Oil Temperature",
        "Coolant Temperature"
    ],
    "Value": [
        sample_vehicle["engine_rpm"],
        sample_vehicle["lub_oil_pressure"],
        sample_vehicle["fuel_pressure"],
        sample_vehicle["coolant_pressure"],
        sample_vehicle["lub_oil_temp"],
        sample_vehicle["coolant_temp"]
    ],
    "Unit": [
        "RPM",
        "bar",
        "bar",
        "bar",
        "°C",
        "°C"
    ]
})

print()
print("=" * 75)
print("SAMPLE VEHICLE SENSOR READINGS")
print("=" * 75)

display(sensor_values)


# ================================================================
# 33. PREDICTION EXAMPLES FROM TEST DATA
# ================================================================

print()
print("=" * 75)
print("SAMPLE MODEL PREDICTIONS")
print("=" * 75)

sample_predictions = X_test.copy()

sample_predictions["Actual Condition"] = y_test.values

sample_predictions["Failure Probability"] = (
    deployed_probabilities
)

sample_predictions["Predicted Condition"] = (
    (deployed_probabilities >= 0.50).astype(int)
)

sample_predictions["Health Score"] = (
    (1 - deployed_probabilities) * 100
)

sample_predictions["Status"] = pd.cut(
    deployed_probabilities,
    bins=[
        -0.01,
        0.25,
        0.50,
        0.75,
        1.01
    ],
    labels=[
        "NORMAL",
        "MONITOR",
        "WARNING",
        "CRITICAL"
    ]
)

display(
    sample_predictions.head(15).round(3)
)


# ================================================================
# 34. SAVE PROJECT DATA AND RESULTS
# ================================================================

print()
print("=" * 75)
print("SAVING PROJECT FILES")
print("=" * 75)

output_directory = "predictive_maintenance_project"

os.makedirs(
    output_directory,
    exist_ok=True
)


# Save cleaned dataset
cleaned_dataset_path = os.path.join(
    output_directory,
    "cleaned_engine_sensor_dataset.csv"
)

df.to_csv(
    cleaned_dataset_path,
    index=False
)


# Save model comparison
comparison_path = os.path.join(
    output_directory,
    "model_comparison.csv"
)

comparison.to_csv(
    comparison_path
)


# Save feature importance
importance_path = os.path.join(
    output_directory,
    "xgboost_feature_importance.csv"
)

xgb_importance.to_csv(
    importance_path,
    index=False
)


# Save SHAP importance
shap_path = os.path.join(
    output_directory,
    "shap_feature_importance.csv"
)

shap_importance.to_csv(
    shap_path,
    index=False
)


# Save sample predictions
prediction_path = os.path.join(
    output_directory,
    "sample_predictions.csv"
)

sample_predictions.to_csv(
    prediction_path,
    index=False
)


# Save models
rf_model_path = os.path.join(
    output_directory,
    "random_forest_model.pkl"
)

xgb_model_path = os.path.join(
    output_directory,
    "xgboost_model.pkl"
)

joblib.dump(
    rf_model,
    rf_model_path
)

joblib.dump(
    xgb_model,
    xgb_model_path
)


# Save scaler information / feature names
metadata = {
    "dataset": "jskswamy/predictive-maintenance-data",
    "features": feature_columns,
    "target": target_column,
    "target_definition": {
        "0": "Normal / Healthy",
        "1": "Maintenance Required / Faulty"
    },
    "selected_model": deployed_model_name,
    "random_forest_metrics": rf_metrics,
    "xgboost_metrics": xgb_metrics
}

metadata_path = os.path.join(
    output_directory,
    "project_metadata.json"
)

with open(
    metadata_path,
    "w"
) as f:

    json.dump(
        metadata,
        f,
        indent=4
    )


print()
print("Project files saved successfully!")
print()

for filename in os.listdir(output_directory):
    print("✔", filename)


# ================================================================
# 35. FINAL PROJECT SUMMARY
# ================================================================

print()
print()
print("=" * 75)
print(" FINAL PROJECT SUMMARY")
print("=" * 75)

print()

print("Dataset:")
print("Real automotive engine sensor dataset")

print()

print("Input Parameters:")
for feature in feature_columns:
    print("•", feature)

print()

print("Target:")
print("Engine Condition")

print()

print("Models:")
print("• Random Forest")
print("• XGBoost")

print()

print("Explainability:")
print("• SHAP")

print()

print("Selected Prediction Model:")
print("•", deployed_model_name)

print()

print("Best Model ROC-AUC:")

if xgb_auc >= rf_auc:
    print(f"• XGBoost: {xgb_auc:.4f}")
else:
    print(f"• Random Forest: {rf_auc:.4f}")

print()

print("Vehicle Health Monitoring:")
print("• Failure Probability")
print("• Vehicle Health Score")
print("• Vehicle Status")
print("• Maintenance Alert")
print("• Major Contributing Factors")

print()

print("Project Pipeline:")
print(
    "Vehicle Sensors"
    " → Data Preprocessing"
    " → ML Model"
    " → Failure Probability"
    " → Vehicle Health Score"
    " → Maintenance Alert"
)

print()

print("=" * 75)
print(" PROJECT EXECUTION COMPLETED SUCCESSFULLY ")
print("=" * 75)
```
